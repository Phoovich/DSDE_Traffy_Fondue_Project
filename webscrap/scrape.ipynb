```python
import requests
import pandas as pd
from io import StringIO
```

```python
url = "https://publicapi.traffy.in.th/teamchadchart-stat-api/download/bangkok_monthly"
file_name = ['bangkok_2025-02', 'bangkok_2025-03', 'bangkok_2025-04', 'bangkok_2025-05',
             'bangkok_2025-06', 'bangkok_2025-07', 'bangkok_2025-08', 'bangkok_2025-09',
             'bangkok_2025-10', 'bangkok_2025-11']
```

```python
df_list = []

for fname in file_name:
    params = {
        "name": "ภูวิศ พึ่งแผ่นดิน",
        "org": "นิสิตจุฬา",
        "email": "phoovich@gmail.com",
        "purpose": "ทำโปรเจค DSDE",
        "file_name": fname,
    }
    r = requests.get(url, params=params)

    # Response IS CSV → use .text, NOT .json()
    csv_text = r.text

    header = "ticket_id,type,organization"
    idx = csv_text.find(header)
    clean_csv = csv_text[idx:]

    df = pd.read_csv(StringIO(clean_csv))
    df_list.append(df)
```

```python
# 3) Concatenate all
final_df = pd.concat(df_list, ignore_index=True)
print("Combined shape:", final_df.shape)
final_df.to_csv("data.csv", index=False)
```

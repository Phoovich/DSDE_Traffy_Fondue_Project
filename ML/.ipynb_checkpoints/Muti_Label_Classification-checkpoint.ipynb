```python
import pandas as pd
```

```python
df = pd.read_csv('../bangkok_traffy.csv')
```

```python
df.head(3)
```

## Data Prep.

```python
# drop null
df = df.dropna(subset=['type','comment'])
print("null in col type = ",df['type'].isna().sum())
print("null in col comment = ",df['comment'].isna().sum())
```

```python
df = df[['comment','type']]
df.head(3)
```

```python
def parse_brace_list(s):
    s = str(s).strip("{}")
    parts = [x.strip() for x in s.split(",")]
    return [p for p in parts if p]

df['type'] = df['type'].apply(parse_brace_list)
df.head(3)
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.pipeline import Pipeline
from sklearn import set_config
set_config(transform_output='pandas')

df = df.reset_index(drop=True)
mlb = MultiLabelBinarizer()
df_type_array = mlb.fit_transform(df['type'])
df_mlb = pd.DataFrame(df_type_array, columns=mlb.classes_)
df_concat = pd.concat([df['comment'],df_mlb],axis=1)
```

```python
df_concat.to_csv('../bangkok_traffy_prep.csv',index=False)
```

## Visualization

```python
import matplotlib.pyplot as plt
from collections import Counter
plt.rcParams['font.family'] = 'Tahoma'

# Flatten all label lists into a single list
flat_labels = [label for label_list in df['type'] for label in label_list]

# Count label frequencies
label_counts = Counter(flat_labels)
sorted_labels = sorted(label_counts.items(), key=lambda x: x[1], reverse=True)

# Unpack
labels, counts = zip(*sorted_labels)

# Plot
plt.figure(figsize=(14, 6))
bars = plt.bar(labels, counts, color='skyblue', edgecolor='black')

# Add value labels on bars
for bar, count in zip(bars, counts):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1, 
             str(count), ha='center', va='bottom', fontsize=8)

plt.xlabel('Labels', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Label Distribution', fontsize=14)
plt.xticks(rotation=45, ha='right',fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
